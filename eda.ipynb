```python
df.columns = df.columns.str.strip()

print(df.columns.tolist())
```

```python
print("Dataset shape:", df.shape)
print("UTR column exists:", "UTR" in df.columns)

display(df.head(10))

print("\nRemaining Paid to prefixes:")
print(
    df["Transaction Details"]
    .str.contains(r"^\s*Paid to\b", case=False, na=False)
    .sum()
)

print("\nRemaining Received from prefixes:")
print(
    df["Transaction Details"]
    .str.contains(r"^\s*Received from\b", case=False, na=False)
    .sum()
)
```

```python
# Remove UTR column
df = df.drop(columns=["UTR"])

# Remove transaction prefixes
df["Transaction Details"] = (
    df["Transaction Details"]
    .astype("string")
    .str.replace(
        r"^\s*(?:Paid to|Received from)\s+",
        "",
        regex=True,
        case=False
    )
    .str.strip()
)

print(df.head())
print(df.columns.tolist())
```
